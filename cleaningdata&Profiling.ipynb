```python
#libraries 
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns

#####################################Dataset Cleaning##################################################

#Loading the dataset
CustomerData_Df = pd.read_csv("/Users/manesh/Library/CloudStorage/OneDrive-UniversityofPlymouth/Dissertation /Datset/customer_churn_business_dataset.csv")

#Confirm the dataset loaded successfully
CustomerData_Df.head()

#Explore and understand the imported dataset
print(CustomerData_Df.shape)
CustomerData_Df.info()

#Dataset summary statistics 
CustomerData_Df.describe()
CustomerData_Df.describe(include='object')

#Rename and Clean the dataset column names 
CustomerData_Df.columns = CustomerData_Df.columns.str.lower().str.strip().str.replace(" ","_")

#Check the dataset column names  are cleaned successfully
CustomerData_Df.columns

#Check the missing values 
CustomerData_Df.isnull().sum()

#Handling the missing values 
CustomerData_Df['complaint_type'] = CustomerData_Df['complaint_type'].fillna('type_missing')

#Check the duplicate values 
print("the duplicated records : ", CustomerData_Df.duplicated().sum())

#Remove the duplicate values
CustomerData_Df.drop_duplicates(inplace=True)  

print("Recheck the duplicated records : ", CustomerData_Df.duplicated().sum())

#Remove irrelevant Columns 
CustomerData_Df.drop(columns=['customer_id'], inplace=True)

#check the irrelevant Column as been removed successfully
CustomerData_Df.columns

#Check Unique Values in Categorical variables

cat_cols = CustomerData_Df.select_dtypes(include=['object', 'category']).columns 

for col in cat_cols:
    print(col)
    print(CustomerData_Df[col].value_counts())
    print("----------------------------")

#Standardize Categories
CustomerData_Df['gender'] = CustomerData_Df['gender'].str.strip().str.title()
CustomerData_Df['city'] = CustomerData_Df['city'].str.strip().str.title()

#Check the categories standardization is done successfully

for col in cat_cols:
    print(col)
    print(CustomerData_Df[col].value_counts())
    print("----------------------------")

#Check the data types of the dataset 
print("Data types of the dataset: ", CustomerData_Df.dtypes)

#Convert the data types to relevant types 
CustomerData_Df['age'] = pd.to_numeric(CustomerData_Df['age'].astype('Int64'))

#Recheck the data types of the dataset 
print("Data types of the dataset (After conversion): ", CustomerData_Df.dtypes)

#Detecting Outliers 

#Using Boxplot 
numerics = CustomerData_Df.select_dtypes(include=['int64','float64'])

for col in numerics.columns:
    plt.figure(figsize=(5,2))
    plt.boxplot(CustomerData_Df[col])
    plt.title(col)
    plt.show()


#IQR Method
Q1 = CustomerData_Df['monthly_fee'].quantile(0.25)
Q3 = CustomerData_Df['monthly_fee'].quantile(0.75)

IQR = Q3-Q1

lower = Q1-1.5*IQR
upper = Q3+1.5*IQR

CustomerData_Df[(CustomerData_Df['monthly_fee']<lower) | (CustomerData_Df['monthly_fee']>upper)]


#Remove Impossible Values
CustomerData_Df[CustomerData_Df['age']<18]

CustomerData_Df[CustomerData_Df['monthly_fee']<0]

#Check Class Balance
CustomerData_Df['churn'].value_counts()

CustomerData_Df['churn'].value_counts(normalize=True)*100


#Check Categorical Values
for col in cat_cols:
    print(CustomerData_Df[col].unique())

#Save Clean Dataset
CustomerData_Df.to_csv("customer_churn_cleaned.csv", index=False)

#Final Validation
print(CustomerData_Df.info())

print(CustomerData_Df.isnull().sum())

print(CustomerData_Df.duplicated().sum())

print(CustomerData_Df.shape)

####################################Dataset Profiling  Report#####################################

#%pip install ydata-profiling

# importing the library
from ydata_profiling import ProfileReport

# Create report
profile = ProfileReport(
    CustomerData_Df, 
    title="Customer Churn Dataset Profiling Report",
    explorative=True
)

profile.to_file("customer_churn_profiling_report.html")
print("🎉 HTML Report generated successfully!")


####################################Dataset Profiling  Report#####################################

#Creating histograms for each numerical variable
numerical_cols =  CustomerData_Df.select_dtypes(include=['int64','float64']).columns

for col in numerical_cols:
    plt.figure(figsize=(6,4))
    CustomerData_Df[col].hist(bins=30)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.show()

#for Categorical variables 
for col in cat_cols:
    plt.figure(figsize=(6,4))
    CustomerData_Df[col].value_counts().plot(kind='bar')
    plt.title(col)
    plt.ylabel("Count")
    plt.show()

#Bivariate Analysis

#Age vs Churn
sns.boxplot(
    x='churn',
    y='age',
    data=CustomerData_Df
)

plt.show()

#Monthly Fee vs Churn
sns.boxplot(
    x='churn',
    y='monthly_fee',
    data=CustomerData_Df
)

plt.show()


#Tenure vs Churn
sns.boxplot(
    x='churn',
    y='tenure_months',
    data=CustomerData_Df
)

plt.show()


#Gender vs Churn
pd.crosstab(
    CustomerData_Df['gender'],
    CustomerData_Df['churn']
).plot(kind='bar')


#Contract Type vs Churn
pd.crosstab(
    CustomerData_Df['contract_type'],
    CustomerData_Df['churn']
).plot(kind='bar')

#Support Tickets vs Churn
sns.boxplot(
    x='churn',
    y='support_tickets',
    data=CustomerData_Df
)

#CSAT Score vs Churn
sns.boxplot(
    x='churn',
    y='csat_score',
    data=CustomerData_Df
)

#NPS Score vs Churn
sns.boxplot(
    x='churn',
    y='nps_score',
    data=CustomerData_Df
)

#Multivariate Analysis

#Correlation Heatmap

plt.figure(figsize=(14,10))

sns.heatmap(
    CustomerData_Df.corr(numeric_only=True),
    annot=False,
    cmap='coolwarm'
)

plt.show()

#Pair Plot
selected = [
    'age',
    'monthly_fee',
    'tenure_months',
    'support_tickets',
    'churn'
]

sns.pairplot(
    CustomerData_Df[selected],
    hue='churn'
)

#Churn Correlation Ranking
corr = CustomerData_Df.corr(numeric_only=True)

corr['churn'].sort_values(ascending=False)


#####################Feature Engineering #####################################

# Estimated Customer Lifetime Value
CustomerData_Df["estimated_clv"] = (
    CustomerData_Df["monthly_fee"] *
    CustomerData_Df["tenure_months"]
)

# Customer Engagement Score
CustomerData_Df["engagement_score"] = (
    CustomerData_Df["monthly_logins"] * 0.4 +
    CustomerData_Df["weekly_active_days"] * 0.3 +
    CustomerData_Df["features_used"] * 0.3
)


# Customer Support Burden
CustomerData_Df["support_burden"] = (
    CustomerData_Df["support_tickets"] +
    CustomerData_Df["escalations"]
)

# Marketing Engagement
CustomerData_Df["marketing_engagement"] = (
    CustomerData_Df["email_open_rate"] *
    CustomerData_Df["marketing_click_rate"]
)

# Revenue per Login
CustomerData_Df["revenue_per_login"] = (
    CustomerData_Df["monthly_fee"] /
    (CustomerData_Df["monthly_logins"] + 1)
)

# Complaint Severity
CustomerData_Df["complaint_severity"] = (
    CustomerData_Df["support_tickets"] *
    CustomerData_Df["avg_resolution_time"]
)

# Customer Activity Index
CustomerData_Df["activity_index"] = (
    CustomerData_Df["weekly_active_days"] *
    CustomerData_Df["avg_session_time"]
)

# Standardise values
CustomerData_Df["price_increase_last_3m"] = (
    CustomerData_Df["price_increase_last_3m"]
        .str.strip()
        .str.title()
)

# Check values
print(CustomerData_Df["price_increase_last_3m"].unique())

# Payment Risk Score
CustomerData_Df["payment_risk"] = (
    CustomerData_Df["payment_failures"] +
    CustomerData_Df["price_increase_last_3m"].map({
        "Yes": 1,
        "No": 0
    })
)

CustomerData_Df["customer_satisfaction_index"] = (
    CustomerData_Df["csat_score"] +
    CustomerData_Df["nps_score"]
) / 2


#Check dataset 
CustomerData_Df.head()

CustomerData_Df.describe()

CustomerData_Df.info()

new_features = [
    "estimated_clv",
    "engagement_score",
    "support_burden",
    "marketing_engagement",
    "revenue_per_login",
    "complaint_severity",
    "activity_index",
    "payment_risk",
    "customer_satisfaction_index"
]

print(CustomerData_Df[new_features].head())
print(CustomerData_Df[new_features].describe())
print(CustomerData_Df[new_features].isnull().sum())

```
